```python
import torch
from torch.utils.data import Dataset,DataLoader
```

```python
class MyDataset(Dataset):
    #初始化
    def __init__(self,data):
        self.data = data
    #数据集大小
    def __len__(self):
        return len(self.data)
    #按索引号获取数据
    def __getitem__(self, idx):
        return self.data[idx]
```

```python
#创建数据集
dataset = MyDataset([2,4,6,7,9])
```
