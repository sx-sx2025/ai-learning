```python
import  torch
```

```python
x = torch.tensor(10.0)
y = torch.tensor(3.0)
```

```python
#定义权重和偏置参数
w = torch.tensor(0.5,requires_grad=True)
b = torch.tensor(1.2,requires_grad=True)
```

```python
#做计算，得到预测输出（前向传播）
z = x * w + b
print(z)
```

```python
#计算Loss
loss_fn = torch.nn.MSELoss()
loss = loss_fn(z, y)
print(loss)
```

```python
#反向传播
loss.backward()
```

```python
#乍看梯度
print(w.grad)
print(b.grad)
```

```python
#乍看叶子节点
print(x.is_leaf)
print(y.is_leaf)
print(w.is_leaf)
print(b.is_leaf)
print(z.is_leaf)
print(loss.is_leaf)
```

```python
print(x.requires_grad)
print(y.requires_grad)
print(w.requires_grad)
print(b.requires_grad)
```
