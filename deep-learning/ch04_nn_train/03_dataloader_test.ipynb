```python
import torch
from torch.utils.data import Dataset, DataLoader, dataloader, dataset
```

```python
#定义数据
X = torch.randint(0,10,(10,3))
y = torch.rand(10)
print(X)
print(y)
```

```python
#创建数据集
dataset = dataset.TensorDataset(X, y)
print(dataset[0])
```

```python
#创建数据加载器
dataloader = DataLoader(dataset,batch_size=4)
```

```python
for x_batch,y_batch in dataloader:
    print(x_batch)
    print(y_batch)
```
