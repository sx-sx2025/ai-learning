```python
import torch
from Demos.win32cred_demo import target
from torch import nn
```

```python
#定义模型输出数据与目标值
output = torch.randn(5)
target = torch.randn(5)
print(output)
print(target)
```

```python
#MAE
loss_fn = nn.L1Loss()
loss = loss_fn(input=output, target=target)
print(loss)
```

```python
#MSE
loss_fn = nn.MSELoss()
loss = loss_fn(input=output, target=target)
print(loss)
```

```python
#平滑L1
loss_fn = nn.SmoothL1Loss()
loss = loss_fn(input=output, target=target)
print(loss)
```
