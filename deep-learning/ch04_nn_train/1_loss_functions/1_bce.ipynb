```
import torch
from debugpy.launcher import output
from torch import nn
```

```
#定义BCE损失函数
loss_fn = nn.BCELoss()
```

```
#模型的输出
output = torch.randn(2,3)
print(output)
```

```
#转化为概率
output_proba = torch.sigmoid(output)
print(output_proba)
```

```
#定义目标值
target = torch.tensor([0,0,1,0,1]).float()
```

```
#计算损失
loss = loss_fn(input=output_proba, target=target)
print(loss)
```
