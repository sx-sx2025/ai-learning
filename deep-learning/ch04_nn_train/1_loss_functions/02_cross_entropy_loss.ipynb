```python
import torch
from Demos.win32cred_demo import target
from torch import nn
```

```python
#多分类损失函数
loss_fn = nn.CrossEntropyLoss()
```

```python
#定义模型输出的预测值，8条数据，6条分类
logits = torch.randn(8,6)
print(logits)
```

```python
target = torch.randint(0,6,(8,))
print(target)
```

```python
loss = loss_fn(logits, target)
print(loss)
```

```python
target_onehot = torch.zeros(8,6)
target_onehot[ torch.arange(8),target ] = 1
print(target_onehot)
```

```python
loss = loss_fn(logits, target_onehot)
print(loss)
```
