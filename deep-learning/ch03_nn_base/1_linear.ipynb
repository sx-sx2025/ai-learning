```python
import torch
print(torch.cuda.is_available())          # 应该为 True
print(torch.cuda.get_device_name(0))     # 应该显示你的 50 系显卡型号
x = torch.ones(2, device='cuda')
print(x + x)                              # 不报错即表示内核可用
```

```python
import torch
print("PyTorch 版本:", torch.__version__)
print("CUDA 可用:", torch.cuda.is_available())

# 关键：查看 PyTorch 编译时支持哪些 GPU 架构
print("支持的 CUDA 架构列表:", torch.cuda.get_arch_list())

# 再测一次最简单的 GPU 运算
x = torch.ones(2, device='cuda')
print("运算结果:", x + x)
```
