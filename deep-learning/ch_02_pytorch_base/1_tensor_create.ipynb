```python

import numpy as np
import torch
from sympy.tensor import tensor

```

```python
tensor1 = torch.tensor(10)
print(tensor1)
```

```python
tensor1 = torch.Tensor(3,4)
print(tensor1)
```

```python
tensor1 = torch.tensor([1,2,3],dtype=torch.int32)
tensor2 = torch.IntTensor(2,3)

print(tensor1.dtype)
print(tensor2.dtype)
```

```python
tensor1 = torch.tensor([1,2,3],dtype=torch.int64)
tensor2 = torch.LongTensor(2,3)

print(tensor1.dtype)
print(tensor2.dtype)
```

```python
tensor1 = torch.linspace(10,30,5)
print(tensor1)
```

```python
tensor1 = np.linspace(10,30,5,endpoint=False)
print(tensor1)
```
